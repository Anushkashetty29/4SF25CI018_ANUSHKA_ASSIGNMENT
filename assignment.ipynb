*5.Create a Pandas DataFrame with 2 columns: Category and Value. Fill the Category column with random categories (A, B, C) and Value with random integers. Group the DataFrame by Category and calculate the sum and mean of Value.*

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(10, 100, size=10)
})

print(df)
```

```python
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("Sum and Mean of each Category:")
print(result)
```

*6.Create a Pandas DataFrame with 3 columns: Product, Category, and Sales. Fill the DataFrame with random data. Group the DataFrame by Category and calculate the total sales for each category.*

```python
df = pd.DataFrame({
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Keyboard', 'Mouse', 'Monitor'],
    'Category': np.random.choice(['Electronics', 'Accessories'], size=6),
    'Sales': np.random.randint(100, 1000, size=6)
})

print(df)
```

```python
result = df.groupby('Category')['Sales'].sum()

print("Total Sales for each Category:")
print(result)
```

*7.Create two Pandas DataFrames with a common column and merge them using the common column.*

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Anu', 'Riya', 'Sam', 'John']
})

print("First DataFrame:")
print(df1)
```

```python
df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [85, 90, 78, 95]
})

print("Second DataFrame:")
print(df2)
```

```python
result = pd.merge(df1, df2, on='ID')

print("Merged DataFrame:")
print(result)
```

*8.Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform basic indexing and slicing operations on the MultiIndex DataFrame.*

```python
arrays = [
    ['A', 'A', 'B', 'B', 'C', 'C'],
    ['One', 'Two', 'One', 'Two', 'One', 'Two']
]

index = pd.MultiIndex.from_arrays(arrays, names=['Category', 'SubCategory'])

df = pd.DataFrame({
    'Value': [10, 20, 30, 40, 50, 60]
}, index=index)

print(df)
```

```python
print(df.loc['A'])
```

```python
print(df.loc[('B', 'One')])
```

```python
print(df.loc['A':'B'])
```

*9.Create a Pandas DataFrame with MultiIndex consisting of Category and SubCategory. Fill the DataFrame with random data and compute the sum of values for each Category and SubCategory.*

```python
arrays = [
    ['A', 'A', 'A', 'B', 'B', 'B'],
    ['One', 'Two', 'One', 'One', 'Two', 'Two']
]

index = pd.MultiIndex.from_arrays(arrays, names=['Category', 'SubCategory'])

df = pd.DataFrame({
    'Value': np.random.randint(10, 100, size=6)
}, index=index)

print(df)
```

```python
result = df.groupby(level=['Category', 'SubCategory']).sum()

print("Sum of values for each Category and SubCategory:")
print(result)
```

*10.Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.*

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)), columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)
```

```python
result = df.apply(lambda x: x * 2)

print("DataFrame after doubling the values:")
print(result)
```

*11.Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.*

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(6, 3)), columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)
```

```python
df['Total'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)

print("DataFrame after adding Total column:")
print(df)
```
